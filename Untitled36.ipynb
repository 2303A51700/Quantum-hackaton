```python
"""
HHL least-squares quantum SVM (P01) -- statevector simulation of the full circuit.

Circuit (qubits: c | flag | clock(nc) | sys(ns)):
  H(c)
  c-multiplexed state prep:  c=0 -> |b>   (b = (0, y)/sqrt(n), padded)
                             c=1 -> |q(x)> (q = (1, k(x_1,x),..,k(x_n,x))/norm, padded)
  QPE:  H^nc on clock; controlled-U^(2^j), U = exp(i F t); inverse QFT
  Eigenvalue inversion: c=0 -> controlled-Ry(2 asin(C/lam~)) on flag (signed lam~);
                        c=1 -> X on flag   (QPE . QPE^dagger = identity on that branch)
  QPE^dagger (QFT, controlled-U^-(2^j), H^nc)
  H(c); measure c and flag.  Post-select flag=1.
  P(c=1 | flag=1) = 1/2 - Re<q|x_hat>*sqrt(p)/(1+p)   ->  sign gives sign f(x).

The solution of F [b; alpha] = [0; y] is never computed classically inside the
quantum path. Classical numpy is used only for (i) building the unitary oracle
exp(iFt) (what Hamiltonian simulation would implement), (ii) choosing t, C from
the spectrum (standard HHL assumption: ||F|| and kappa known), and
(iii) reference quantities for the fidelity / residual metrics.
"""
import time, json, sys
import numpy as np
from scipy.linalg import expm, hadamard
from sklearn.datasets import load_iris, load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score,
                             confusion_matrix, roc_auc_score)


# ----------------------------------------------------------------- data
def load_binary(name):
    if name == "iris":          # versicolor (1) vs virginica (2): the non-trivial pair
        d = load_iris(); keep = (1, 2)
    elif name == "wine":        # cultivar 0 vs 1
        d = load_wine(); keep = (0, 1)
    m = np.isin(d.target, keep)
    X, y = d.data[m], np.where(d.target[m] == keep[1], 1, -1)
    return X, y


def rbf(A, B, sig2):
    d2 = ((A[:, None, :] - B[None, :, :]) ** 2).sum(-1)
    return np.exp(-d2 / (2 * sig2))


# ----------------------------------------------------------------- HHL
def build_system(Xtr, ytr, gamma, sig2):
    n = len(ytr)
    K = rbf(Xtr, Xtr, sig2)
    F = np.zeros((n + 1, n + 1))
    F[0, 1:] = 1; F[1:, 0] = 1
    F[1:, 1:] = K + np.eye(n) / gamma
    rhs = np.concatenate([[0.0], ytr.astype(float)])
    ns = int(np.ceil(np.log2(n + 1)))
    M = 2 ** ns
    Fp = np.eye(M) * (1 + 1 / gamma)          # pad block: spectrum-neutral diagonal
    Fp[:n + 1, :n + 1] = F
    bp = np.zeros(M); bp[:n + 1] = rhs
    return Fp, bp / np.linalg.norm(bp), np.linalg.norm(bp), ns, n


def hhl_circuit(F, b, nc, C_frac=0.95):
    """Simulate the HHL branch of the circuit (c=0). Returns amplitudes [flag, clock, sys]."""
    M = F.shape[0]; N = 2 ** nc
    ev = np.linalg.eigvalsh(F)                 # parameter setting only (t, C), not the solve
    lam_max, lam_min = np.abs(ev).max(), np.abs(ev).min()
    t = 2 * np.pi * 0.49 / lam_max
    U = expm(1j * F * t)
    P = [U]
    for _ in range(nc - 1):
        P.append(P[-1] @ P[-1])                 # U^(2^j) by repeated squaring
    ms = np.arange(N)
    psi = np.tile(b.astype(complex), (N, 1)) / np.sqrt(N)         # H^nc on clock, |b> on sys
    for j in range(nc):                                           # controlled-U^(2^j)
        mk = ((ms >> j) & 1) == 1
        psi[mk] = psi[mk] @ P[j].T
    W = np.exp(-2j * np.pi * np.outer(ms, ms) / N) / np.sqrt(N)   # inverse QFT
    psi = W @ psi
    ks = np.where(ms >= N // 2, ms - N, ms)
    lam_t = np.zeros(N); lam_t[ks != 0] = 2 * np.pi * ks[ks != 0] / (t * N)
    C = C_frac * lam_min
    amp = np.zeros(N)
    nz = ks != 0
    amp[nz] = np.clip(C / lam_t[nz], -1, 1)
    out = np.zeros((2, N, M), complex)
    out[0] = psi * np.sqrt(1 - amp ** 2)[:, None]
    out[1] = psi * amp[:, None]
    for f in range(2):                                            # QPE^dagger
        s = W.conj().T @ out[f]
        for j in reversed(range(nc)):
            mk = ((ms >> j) & 1) == 1
            s[mk] = s[mk] @ P[j].conj()          # (U^-k) = conj(U^k)^T acting on row vecs
        Hn = hadamard(N) / np.sqrt(N)
        out[f] = Hn @ s
    return out, dict(t=t, C=C, kappa=lam_max / lam_min, lam_min=lam_min, lam_max=lam_max)


def cnot_qsd(n):                                # CNOT count of generic n-qubit unitary (QSD)
    return 23 / 48 * 4 ** n - 1.5 * 2 ** n + 4 / 3


def cost_model(ns, nc):
    cU = nc * cnot_qsd(ns + 1)                  # controlled-U^(2^j), each a dense (ns+1)-qubit unitary
    qft = nc * (nc - 1)                         # nc(nc-1)/2 CPhase * 2 CNOT
    prep = 2 ** (ns + 1)                        # multiplexed amplitude loading (1 ctrl + ns)
    rot = 2 ** nc                               # uniformly-controlled Ry, nc controls
    total = 2 * (cU + qft) + prep + rot
    return dict(qubits=ns + nc + 2, two_qubit_gates_est=int(total),
                depth_est=int(total))           # serial on shared sys qubits: depth ~ 2q-gate count


# ----------------------------------------------------------------- experiment
def run(name, seeds, nc, gamma, shots, n_per_class=10):
    X, y = load_binary(name)
    d = X.shape[1]
    sig2 = float(d)
    rows, cms, aucs_all = [], [], []
    for seed in seeds:
        rng = np.random.default_rng(seed)
        tr = np.concatenate([rng.choice(np.where(y == c)[0], n_per_class, replace=False) for c in (-1, 1)])
        te = np.setdiff1d(np.arange(len(y)), tr)
        sc = StandardScaler().fit(X[tr])
        Xtr, Xte = sc.transform(X[tr]), sc.transform(X[te])
        ytr, yte = y[tr], y[te]

        # ---- training = build system + run HHL
        t0 = time.time()
        F, b, bnorm, ns, n = build_system(Xtr, ytr, gamma, sig2)
        out, info = hhl_circuit(F, b, nc)
        t_train = time.time() - t0
        p = float((np.abs(out[1]) ** 2).sum())
        v = out[1][0]                                   # flag=1, clock=|0..0>
        # diagnostics (statevector access; reference solve only for metrics)
        x_ref = np.linalg.solve(F, b)
        xr = x_ref / np.linalg.norm(x_ref)
        fid = float((np.abs(out[1] @ xr) ** 2).sum() / p)       # sum over clock of |<x|psi_c>|^2
        x_hat = np.real(v) / info["C"]
        resid = float(np.linalg.norm(F @ x_hat - b) / np.linalg.norm(b))

        # ---- inference
        t0 = time.time()
        Kte = rbf(Xte, Xtr, sig2)
        Q = np.zeros((len(te), F.shape[0])); Q[:, 0] = 1; Q[:, 1:n + 1] = Kte
        Q /= np.linalg.norm(Q, axis=1, keepdims=True)
        cross = Q @ np.real(v)                          # Re<q,0,1|Psi0>
        P1 = np.clip((p + 1 - 2 * cross) / (2 * (p + 1)), 0, 1)
        n1 = rng.binomial(shots, (p + 1) / 2, size=len(te))
        m1 = rng.binomial(n1, P1)
        score_shot = 0.5 - m1 / np.maximum(n1, 1)
        score_ideal = 0.5 - P1
        tie = score_shot == 0
        pred = np.where(score_shot > 0, 1, -1)
        pred[tie] = rng.choice([-1, 1], tie.sum())
        pred_ideal = np.where(score_ideal > 0, 1, -1)
        t_inf = time.time() - t0

        # classical LS-SVM baseline (NOT HHL)
        sol = np.linalg.solve(F[:n + 1, :n + 1], np.concatenate([[0], ytr.astype(float)]))
        fb = np.concatenate([np.ones((len(te), 1)), Kte], 1) @ sol
        pred_cl = np.where(fb > 0, 1, -1)

        def met(pr, sc_):
            return dict(acc=accuracy_score(yte, pr), bacc=balanced_accuracy_score(yte, pr),
                        f1=f1_score(yte, pr, average="macro"), auc=roc_auc_score(yte, sc_))
        r = dict(seed=seed, ns=ns, kappa=info["kappa"], p_succ=p, fidelity=fid, residual=resid,
                 train_s=t_train, infer_s=t_inf, n_test=len(te))
        for tag, (pr, sc_) in dict(shots=(pred, score_shot), ideal=(pred_ideal, score_ideal),
                                   classical=(pred_cl, fb)).items():
            for k, vv in met(pr, sc_).items():
                r[f"{tag}_{k}"] = vv
        rows.append(r)
        cms.append(confusion_matrix(yte, pred, labels=[-1, 1]))
    cost = cost_model(rows[0]["ns"], nc)
    cost.update(shots_per_test_point=shots,
                circuit_evals_per_seed=int(rows[0]["n_test"] * shots),
                clock_qubits=nc, sys_qubits=rows[0]["ns"], train_samples=2 * n_per_class,
                padded_dim=2 ** rows[0]["ns"])
    return rows, np.sum(cms, 0), cost


if __name__ == "__main__":
    # Use defaults if running in Colab/Jupyter where sys.argv contains kernel arguments
    nc = 9
    gamma = 2.0
    shots = 20000

    if len(sys.argv) > 1 and not sys.argv[1].startswith('-f'):
        try:
            nc = int(sys.argv[1])
            if len(sys.argv) > 2:
                gamma = float(sys.argv[2])
            if len(sys.argv) > 3:
                shots = int(sys.argv[3])
        except ValueError:
            pass

    seeds = list(range(5))
    res = {}
    for name in ("iris", "wine"):
        rows, cm, cost = run(name, seeds, nc, gamma, shots)
        res[name] = dict(rows=rows, confusion_sum=cm.tolist(), cost=cost)
        keys = [k for k in rows[0] if k not in ("seed",)]
        print(f"\n== {name} ==")
        for k in keys:
            a = np.array([r[k] for r in rows], float)
            print(f"{k:22s} {a.mean():.4f} +- {a.std():.4f}")
        print("confusion (sum over seeds, rows=true[-1,+1], cols=pred):", cm.tolist())
        print("cost:", cost)
    json.dump(res, open("results.json", "w"), indent=1, default=float)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Retrieve data from kernel state variable 'res'
try:
    data = res
except NameError:
    import json
    with open('results.json', 'r') as f:
        data = json.load(f)

datasets = list(data.keys())
metrics = ['acc', 'f1', 'auc']
modes = ['shots', 'ideal', 'classical']
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']

fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharey=True)

for idx, name in enumerate(datasets):
    rows = data[name]['rows']

    # Calculate mean and std for each mode and metric
    means = {mode: [] for mode in modes}
    stds = {mode: [] for mode in modes}

    for mode in modes:
        for metric in metrics:
            values = [r[f'{mode}_{metric}'] for r in rows]
            means[mode].append(np.mean(values))
            stds[mode].append(np.std(values))

    ax = axes[idx]
    x = np.arange(len(metrics))
    width = 0.25

    for m_idx, mode in enumerate(modes):
        ax.bar(x + (m_idx - 1) * width, means[mode], width, yerr=stds[mode],
               label=mode.capitalize(), color=colors[m_idx], capsize=5, alpha=0.85)

    ax.set_title(f'{name.capitalize()} Dataset Metrics', fontsize=14, fontweight='bold')
    ax.set_xticks(x)
    ax.set_xticklabels([m.upper() if m != 'acc' else 'Accuracy' for m in metrics], fontsize=12)
    ax.set_ylim(0.8, 1.02)
    ax.grid(axis='y', linestyle='--', alpha=0.5)
    if idx == 0:
        ax.set_ylabel('Score', fontsize=12)
        ax.legend(loc='lower left', fontsize=11)

plt.tight_layout()
plt.show()
```
