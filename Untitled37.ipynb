```python
#!/usr/bin/env python3
"""
P01 - HHL least-squares quantum SVM (LS-QSVM) on binary subsets of Iris (D01) and Wine (D02).

What this file does
-------------------
* Builds the LS-SVM linear system  F [b; alpha] = [0; y]  and pads it to a power of two
  (identity block on the padding, zeros in the right-hand side, so the solution is unchanged).
* Runs a REAL HHL circuit on a statevector simulator written in NumPy:
      amplitude-encoded state preparation of |y>
      quantum phase estimation (H on clock, controlled-U^(2^j), inverse QFT)
      eigenvalue-inversion rotation on an ancilla (signed eigenvalues, two's-complement clock)
      uncomputation of the phase estimation
      post-selection on ancilla = 1 and clock = 0  (success event)
  No classical solver is used to produce the solution state or the predictions.
* Readout: signed overlap <v|x> between the post-selected solution state |x> and the query
  vector v(q) = [1, k(x_1,q), ..., k(x_M,q), 0...] via an interference (Hadamard-test style)
  measurement on a readout qubit, with finite-shot sampling. sign(<v|x>) is the class.
* Classical `np.linalg.solve` is used ONLY as an evaluation reference for solution-state
  fidelity and residual. It never feeds the predictions.

Simulator notes (be honest about these in the report)
-----------------------------------------------------
* The controlled-U^(2^j) = exp(i*F*t*2^j) gates are built with scipy.linalg.expm as dense unitaries
  (exact Hamiltonian simulation). On hardware these must be compiled (Trotter, qubitization, or
  unitary synthesis); the depth reported here is a logical estimate, not a transpiled depth.
* The training-side HHL state does not depend on the query, so it is simulated once and reused.
  On hardware the circuit is re-run for every shot of every query.

Usage
-----
    python hhl_qsvm.py --dataset iris --classes 1 2 --m-train 14 --clock 6 --shots 8192
    python hhl_qsvm.py --dataset wine --classes 0 1 --seeds 5
    python hhl_qsvm.py --dataset iris --classes 1 2 --shots 0     # exact (infinite-shot) readout
"""
import argparse
import json
import sys
import time
from math import ceil, log2

import numpy as np
from scipy.linalg import expm
from sklearn.datasets import load_iris, load_wine
from sklearn.decomposition import PCA
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, confusion_matrix,
                             f1_score, roc_auc_score)
from sklearn.preprocessing import StandardScaler


# --------------------------------------------------------------------------- data
def load_binary(name, classes):
    data = load_iris() if name == "iris" else load_wine()
    X, y = data.data, data.target
    mask = np.isin(y, classes)
    X, y = X[mask], y[mask]
    y = np.where(y == classes[1], 1, -1)
    return X, y


def kernel(A, B, kind="linear", gamma_rbf=0.5):
    if kind == "linear":
        return A @ B.T
    d2 = ((A[:, None, :] - B[None, :, :]) ** 2).sum(-1)
    return np.exp(-gamma_rbf * d2)


# --------------------------------------------------------------------------- LS-SVM system
def build_ls_svm_system(K, y, gamma):
    """F = [[0, 1^T],[1, K + I/gamma]],  rhs = [0; y], padded to the next power of two."""
    M = len(y)
    F = np.zeros((M + 1, M + 1))
    F[0, 1:] = 1.0
    F[1:, 0] = 1.0
    F[1:, 1:] = K + np.eye(M) / gamma
    rhs = np.concatenate([[0.0], y.astype(float)])
    n = max(1, ceil(log2(M + 1)))
    N = 2 ** n
    Fp = np.eye(N)                      # padding: identity on the extra diagonal
    Fp[: M + 1, : M + 1] = F
    rp = np.zeros(N)                    # padding: zeros in the right-hand side
    rp[: M + 1] = rhs
    return Fp, rp, n, M + 1


# --------------------------------------------------------------------------- HHL simulator
class HHL:
    """Statevector register layout: psi[anc, clock, system]."""

    def __init__(self, F, rhs, clock_qubits):
        self.N = F.shape[0]
        self.n = int(log2(self.N))
        self.m = clock_qubits
        self.Nc = 2 ** self.m
        self.F = F
        self.rhs = rhs
        # Classical preprocessing: scale so all |eigenvalues| <= 1 (spectral norm bound).
        self.nu = np.linalg.norm(F, 2)
        Fs = F / self.nu
        # Choose t so lambda = 1 maps to the largest positive signed clock value Nc/2 - 1.
        self.kmax = self.Nc // 2 - 1
        self.t = 2 * np.pi * self.kmax / self.Nc
        U = expm(1j * Fs * self.t)
        self.U_pows = [U]
        for _ in range(1, self.m):
            self.U_pows.append(self.U_pows[-1] @ self.U_pows[-1])
        c = np.arange(self.Nc)
        self.bits = [((c >> j) & 1).astype(bool) for j in range(self.m)]
        # Hadamard^m and inverse QFT as dense unitaries on the clock register
        H1 = np.array([[1, 1], [1, -1]]) / np.sqrt(2)
        Hm = np.array([[1.0]])
        for _ in range(self.m):
            Hm = np.kron(Hm, H1)
        self.Hm = Hm
        self.IQFT = np.exp(-2j * np.pi * np.outer(c, c) / self.Nc) / np.sqrt(self.Nc)
        # Signed clock value (two's complement) and rotation angle: sin(theta/2) = 1/k, 0 for k = 0
        ks = np.where(c >= self.Nc // 2, c - self.Nc, c)
        self.ks = ks
        self.theta = np.zeros(self.Nc)
        nz = ks != 0
        self.theta[nz] = 2 * np.arcsin(1.0 / ks[nz])

    # --- circuit blocks
    def _ctrl_u(self, psi, dagger=False):
        for j in range(self.m):
            Uj = self.U_pows[j].conj().T if dagger else self.U_pows[j]
            psi[:, self.bits[j], :] = psi[:, self.bits[j], :] @ Uj.T
        return psi

    def _qpe(self, psi):
        psi = np.einsum("ck,aks->acs", self.Hm, psi)
        psi = self._ctrl_u(psi)
        return np.einsum("kc,acs->aks", self.IQFT, psi)

    def _qpe_inverse(self, psi):
        psi = np.einsum("kc,acs->aks", self.IQFT.conj().T, psi)
        psi = self._ctrl_u(psi, dagger=True)
        return np.einsum("ck,aks->acs", self.Hm.conj().T, psi)

    def _eigenvalue_inversion(self, psi):
        cth, sth = np.cos(self.theta / 2)[None, :, None], np.sin(self.theta / 2)[None, :, None]
        out = np.empty_like(psi)
        out[0] = cth[0] * psi[0] - sth[0] * psi[1]
        out[1] = sth[0] * psi[0] + cth[0] * psi[1]
        return out

    # --- full run
    def run(self):
        t0 = time.perf_counter()
        psi = np.zeros((2, self.Nc, self.N), dtype=complex)
        psi[0, 0, :] = self.rhs / np.linalg.norm(self.rhs)      # state preparation |y>
        psi = self._qpe(psi)
        psi = self._eigenvalue_inversion(psi)
        psi = self._qpe_inverse(psi)
        w = psi[1, 0, :]                                         # ancilla = 1, clock = 0
        self.p_success = float(np.vdot(w, w).real)
        self.psi_out = psi
        self.w = w
        self.x_hat = w / np.sqrt(self.p_success)
        self.run_time = time.perf_counter() - t0
        return self.x_hat

    # --- evaluation-only diagnostics (use a classical reference; NOT used for prediction)
    def diagnostics(self):
        x_ref = np.linalg.solve(self.F, self.rhs)
        x_ref_hat = x_ref / np.linalg.norm(x_ref)
        overlap = np.vdot(x_ref_hat, self.x_hat)
        fidelity = float(abs(overlap) ** 2)
        Fx = self.F @ self.x_hat
        s = np.vdot(Fx, self.rhs).real / np.vdot(Fx, Fx).real      # best positive-or-not scale
        residual = float(np.linalg.norm(self.F @ (s * self.x_hat) - self.rhs) / np.linalg.norm(self.rhs))
        return {"fidelity": fidelity, "residual": residual, "p_success": self.p_success,
                "kappa": float(np.linalg.cond(self.F)), "scale_sign": float(np.sign(s))}

    # --- readout by interference on a readout qubit r
    # state before post-selection: (|0>_r |HHL out> + |1>_r |anc=1, clock=0, v_hat>) / sqrt(2)
    def readout_probs(self, v):
        """Return (P_post, P_plus): probability the post-selection passes, and P(r = +) given it passes."""
        nv = np.linalg.norm(v)
        v_hat = v / nv
        p = self.p_success
        ov = np.vdot(v_hat, self.w).real                         # <v_hat|w> = sqrt(p) <v_hat|x_hat>
        p_post = (1.0 + p) / 2.0
        p_plus = 0.5 * (1.0 + 2.0 * ov / (1.0 + p))
        return p_post, float(np.clip(p_plus, 0.0, 1.0)), nv


def estimate_score(hhl, v, shots, p_hat, rng):
    """Signed score proportional to f(q) = <v|x>. shots=0 -> exact."""
    p_post, p_plus, nv = hhl.readout_probs(v)
    if shots == 0:
        ex = 2 * p_plus - 1.0
        n_post = np.inf
    else:
        n_post = rng.binomial(shots, p_post)
        if n_post == 0:
            return 0.0, 0
        ex = 2 * rng.binomial(n_post, p_plus) / n_post - 1.0
    # E[+1/-1 outcome] = 2 sqrt(p) <v_hat|x_hat> / (1 + p)  ->  invert with the estimated p
    overlap = ex * (1.0 + p_hat) / (2.0 * np.sqrt(p_hat))
    return float(nv * overlap), (n_post if shots else 0)


# --------------------------------------------------------------------------- cost model
def cnot_generic_unitary(q):
    """Shende et al. CNOT count for a generic q-qubit unitary (logical estimate)."""
    return int(ceil(23 / 48 * 4 ** q - 1.5 * 2 ** q + 4 / 3))


def cost_estimate(n, m):
    ctrl_u = 2 * m
    cx_ctrl_u = ctrl_u * cnot_generic_unitary(n + 1)
    cx_prep = 2 ** (n + 1) - 2 * n - 2
    cx_rot = 2 ** m                      # uniformly controlled RY
    cx_qft = 2 * m * (m - 1)             # two (inverse) QFTs, 2 CNOT per controlled phase
    layers = 1 + (1 + m + (2 * m - 1)) + 1 + (1 + m + (2 * m - 1))   # logical, register-level
    return {"qubits_hhl": n + m + 1, "qubits_with_readout": n + m + 2,
            "logical_depth_layers": layers, "controlled_U_calls": ctrl_u,
            "cx_estimate": cx_prep + cx_ctrl_u + cx_rot + cx_qft}


# --------------------------------------------------------------------------- one experiment
def run_once(args, seed):
    rng = np.random.default_rng(seed)
    X, y = load_binary(args.dataset, args.classes)
    idx_pos, idx_neg = np.where(y == 1)[0], np.where(y == -1)[0]
    mp = args.m_train // 2
    mn = args.m_train - mp
    tr = np.concatenate([rng.choice(idx_pos, mp, replace=False), rng.choice(idx_neg, mn, replace=False)])
    te = np.setdiff1d(np.arange(len(y)), tr)
    Xtr, ytr, Xte, yte = X[tr], y[tr], X[te], y[te]
    sc = StandardScaler().fit(Xtr)
    Xtr, Xte = sc.transform(Xtr), sc.transform(Xte)
    if args.pca:
        pca = PCA(args.pca).fit(Xtr)
        Xtr, Xte = pca.transform(Xtr), pca.transform(Xte)

    # ---- training: build system, run HHL
    t_train0 = time.perf_counter()
    K = kernel(Xtr, Xtr, args.kernel, args.rbf_gamma)
    F, rhs, n, dim = build_ls_svm_system(K, ytr, args.gamma)
    hhl = HHL(F, rhs, args.clock)
    hhl.run()
    # calibration of p_success (shots counted as training-side cost)
    if args.shots == 0:
        p_hat, cal_shots = hhl.p_success, 0
    else:
        cal_shots = args.cal_shots
        p_hat = max(rng.binomial(cal_shots, hhl.p_success) / cal_shots, 1e-6)
    train_time = time.perf_counter() - t_train0
    diag = hhl.diagnostics()               # evaluation only

    # ---- inference
    t_inf0 = time.perf_counter()
    scores, raw_shots = [], 0
    for q in Xte:
        v = np.zeros(F.shape[0])
        v[0] = 1.0
        v[1: len(ytr) + 1] = kernel(Xtr, q[None, :], args.kernel, args.rbf_gamma)[:, 0]
        s, _ = estimate_score(hhl, v, args.shots, p_hat, rng)
        scores.append(s)
        raw_shots += args.shots
    inf_time = time.perf_counter() - t_inf0
    scores = np.array(scores)
    pred = np.where(scores > 0, 1, -1)

    res = {
        "seed": seed,
        "accuracy": accuracy_score(yte, pred),
        "balanced_accuracy": balanced_accuracy_score(yte, pred),
        "macro_f1": f1_score(yte, pred, average="macro"),
        "auroc": roc_auc_score(yte, scores) if len(np.unique(scores)) > 1 else float("nan"),
        "confusion_matrix": confusion_matrix(yte, pred, labels=[-1, 1]).tolist(),
        "fidelity": diag["fidelity"], "residual": diag["residual"],
        "p_success": diag["p_success"], "kappa": diag["kappa"],
        "n_system_qubits": n, "padded_dim": F.shape[0], "system_dim_before_padding": dim,
        "train_time_s": train_time, "inference_time_total_s": inf_time,
        "inference_time_per_sample_s": inf_time / len(Xte),
        "circuit_evals_train": 1 + (cal_shots > 0),
        "circuit_evals_inference": len(Xte) if args.shots else 0,
        "shots_train_calibration": cal_shots, "shots_inference_total": raw_shots,
        "n_test": len(Xte),
    }
    res.update(cost_estimate(n, args.clock))
    return res


def main():
    ap = argparse.ArgumentParser()
    ap.add_argument("--dataset", choices=["iris", "wine"], default="iris")
    ap.add_argument("--classes", type=int, nargs=2, default=None,
                    help="two class labels (iris default 1 2, wine default 0 1); second = +1")
    ap.add_argument("--m-train", type=int, default=14, help="training samples M (M+1 is padded to 2^n)")
    ap.add_argument("--clock", type=int, default=6, help="clock qubits m (resolution ~ 2^(m-1) >= kappa)")
    ap.add_argument("--gamma", type=float, default=1.0, help="LS-SVM regularisation gamma")
    ap.add_argument("--kernel", choices=["linear", "rbf"], default="linear")
    ap.add_argument("--rbf-gamma", type=float, default=0.5)
    ap.add_argument("--pca", type=int, default=0, help="PCA components (0 = off)")
    ap.add_argument("--shots", type=int, default=8192, help="raw shots per test query (0 = exact readout)")
    ap.add_argument("--cal-shots", type=int, default=20000, help="shots to estimate p_success")
    ap.add_argument("--seeds", type=int, default=5)
    ap.add_argument("--out", default="hhl_qsvm_results.json")

    # Fix for running inside Jupyter/Colab notebooks:
    # If '-f' is passed (used by the Jupyter kernel connection), we parse with empty arguments list.
    if len(sys.argv) > 1 and sys.argv[1] == "-f":
        args = ap.parse_args([])
    else:
        args = ap.parse_args()

    if args.classes is None:
        args.classes = [1, 2] if args.dataset == "iris" else [0, 1]

    results = [run_once(args, s) for s in range(args.seeds)]
    keys = ["accuracy", "balanced_accuracy", "macro_f1", "auroc", "fidelity", "residual", "p_success",
            "train_time_s", "inference_time_per_sample_s"]
    print(f"\n{args.dataset.upper()} classes={args.classes} M={args.m_train} clock={args.clock} "
          f"gamma={args.gamma} kernel={args.kernel} shots={args.shots} seeds={args.seeds}")
    r0 = results[0]
    print(f"system dim {r0['system_dim_before_padding']} -> padded {r0['padded_dim']} "
          f"({r0['n_system_qubits']} system qubits), kappa(seed0)={r0['kappa']:.1f}")
    for k in keys:
        v = np.array([r[k] for r in results], dtype=float)
        print(f"  {k:30s} {np.nanmean(v):.4f} +/- {np.nanstd(v):.4f}")
    cm = np.sum([np.array(r["confusion_matrix"]) for r in results], axis=0)
    print("  confusion matrix summed over seeds (rows true [-1,+1], cols pred [-1,+1]):")
    print("  ", cm.tolist())
    print("  cost:", {k: r0[k] for k in ["qubits_hhl", "qubits_with_readout", "logical_depth_layers",
                                        "controlled_U_calls", "cx_estimate", "circuit_evals_train",
                                        "circuit_evals_inference", "shots_train_calibration",
                                        "shots_inference_total"]})
    with open(args.out, "w") as f:
        json.dump(results, f, indent=2)
    print(f"saved {args.out}")


if __name__ == "__main__":
    main()
```
