```python
"""
Quantum circuit learning (QCL) for function learning -- P08
Datasets: Diabetes regression (D07) and Friedman-1 regression (D08, generated).

Model (Mitarai et al. style, simulated with an exact numpy statevector)
-----------------------------------------------------------------------
* One qubit per feature (n = 10).  Features are min-max scaled to [-1, 1] (scaler fit on the
  fitting partition only, others clipped).  Encoding per qubit: RZ(arccos x^2) RY(arcsin x) |0>.
* Variational part, L layers: RY(theta) on every qubit, RZ(phi) on every qubit, CZ ring.
  P = 2*n*L circuit parameters.
* Readout: one Z-basis measurement of all qubits; prediction = sum_i w_i <Z_i> + b with
  classical weights w, b (all <Z_i> come from the same shots, no extra circuits).
* Training: Adam on mini-batches, MSE loss on standardised targets.
  Gradients are computed with the adjoint method (checked against the parameter-shift rule in
  the self-check).  Circuit calls are COUNTED AS IF parameter-shift were used on hardware:
  per mini-batch step  B * (1 + 2P)  circuits (B forward + 2P shifted circuits per sample),
  validation evaluation  n_val  circuits per epoch.
* Early stopping on a validation split (patience); parameters of the best epoch are restored.
  "Circuit calls to convergence" = calls up to the best-validation epoch.
* Training is simulated with exact expectations; inference is reported both exact and with
  finite shots (multinomial sampling of the joint bitstring distribution).
"""
import json
import math
import sys
import time

import numpy as np
from sklearn.datasets import load_diabetes, make_friedman1
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

N_Q = 10          # qubits (= features)
LAYERS = 3
BATCH = 32
LR = 0.05
MAX_EPOCHS = 60
PATIENCE = 8
INFER_SHOTS = 1024
TRAIN_SHOTS_ASSUMED = 1024   # shots per circuit assumed for the hardware cost figure


def ry(t):
    c, s = math.cos(t / 2), math.sin(t / 2)
    return np.array([[c, -s], [s, c]], dtype=complex)


def rz(t):
    return np.diag([np.exp(-0.5j * t), np.exp(0.5j * t)])


class QCL:
    def __init__(self, n=N_Q, layers=LAYERS):
        self.n, self.L = n, layers
        self.dim = 2 ** n
        self.P = 2 * n * layers
        idx = np.arange(self.dim)
        bits = np.array([(idx >> (n - 1 - i)) & 1 for i in range(n)])  # qubit 0 = most significant
        self.zs = (1 - 2 * bits).astype(float)                           # (n, dim)
        cz = np.ones(self.dim)
        for i in range(n):
            cz *= 1 - 2 * (bits[i] & bits[(i + 1) % n])
        self.cz = cz
        self.ops = []
        for l in range(layers):
            for q in range(n):
                self.ops.append(("ry", q, l * 2 * n + q))
            for q in range(n):
                self.ops.append(("rz", q, l * 2 * n + n + q))
            self.ops.append(("cz", None, None))
        # analytic cost model
        self.depth = 2 + 4 * layers          # encoding (2) + per layer RY, RZ, CZ ring (2 layers)
        self.n_cz = n * layers
        self.n_1q = 2 * n + 2 * n * layers

    # ---- state handling
    def encode(self, X):
        B, n = X.shape
        a, b = np.arcsin(X), np.arccos(X ** 2)
        v0 = np.exp(-0.5j * b) * np.cos(a / 2)
        v1 = np.exp(0.5j * b) * np.sin(a / 2)
        psi = np.ones((B, 1), complex)
        for i in range(n):
            v = np.stack([v0[:, i], v1[:, i]], axis=1)
            psi = (psi[:, :, None] * v[:, None, :]).reshape(B, -1)
        return psi

    def _apply(self, psi, U, q):
        B = psi.shape[0]
        p4 = psi.reshape(B, 2 ** q, 2, 2 ** (self.n - q - 1))
        return np.einsum("ab,xlbr->xlar", U, p4).reshape(B, -1)

    def forward(self, theta, psi0):
        psi = psi0
        for kind, q, k in self.ops:
            if kind == "cz":
                psi = psi * self.cz
            elif kind == "ry":
                psi = self._apply(psi, ry(theta[k]), q)
            else:
                psi = self._apply(psi, rz(theta[k]), q)
        return psi

    def zexp(self, psi):
        return (np.abs(psi) ** 2) @ self.zs.T

    # ---- adjoint gradient of  sum_b r_b * E_b,  E_b = <psi_b| sum_i w_i Z_i |psi_b>
    def grad_theta(self, theta, w, psiN, r):
        Od = w @ self.zs
        lam = psiN * Od[None, :] * r[:, None]
        phi = psiN.copy()
        g = np.zeros(self.P)
        B, n = psiN.shape[0], self.n
        for kind, q, k in reversed(self.ops):
            if kind == "cz":
                phi = phi * self.cz
                lam = lam * self.cz
                continue
            sh = (B, 2 ** q, 2, 2 ** (n - q - 1))
            ph4, lm4 = phi.reshape(sh), lam.reshape(sh)
            if kind == "ry":
                d = np.stack([-0.5 * ph4[:, :, 1, :], 0.5 * ph4[:, :, 0, :]], axis=2)
                U = ry(-theta[k])
            else:
                d = np.stack([-0.5j * ph4[:, :, 0, :], 0.5j * ph4[:, :, 1, :]], axis=2)
                U = rz(-theta[k])
            g[k] += 2.0 * np.real(np.sum(np.conj(lm4) * d))
            phi = self._apply(phi, U, q)
            lam = self._apply(lam, U, q)
        return g


def reg_metrics(y, p):
    res = p - y
    return {"mae": float(np.mean(np.abs(res))), "rmse": float(np.sqrt(np.mean(res ** 2))),
            "r2": float(r2_score(y, p)), "bias": float(np.mean(res))}


def shot_zexp(model, psi, shots, rng):
    probs = np.abs(psi) ** 2
    counts = np.empty_like(probs)
    for i, p in enumerate(probs):
        p = p / p.sum()
        counts[i] = rng.multinomial(shots, p)
    return counts @ model.zs.T / shots


def get_data(name, seed):
    if name == "diabetes_D07":
        X, y = load_diabetes(return_X_y=True)
    else:
        X, y = make_friedman1(n_samples=500, n_features=10, noise=1.0, random_state=seed)
    return X, y


def run_once(name, seed):
    X, y = get_data(name, seed)
    Xtmp, Xte, ytmp, yte = train_test_split(X, y, test_size=0.3, random_state=seed)
    Xf, Xv, yf, yv = train_test_split(Xtmp, ytmp, test_size=0.15, random_state=seed)
    sc = MinMaxScaler((-1, 1)).fit(Xf)                         # fit on fitting partition only
    f = lambda A: np.clip(sc.transform(A), -1, 1)
    Xf_s, Xv_s, Xte_s = f(Xf), f(Xv), f(Xte)
    ym, ys = yf.mean(), yf.std()
    yf_s, yv_s = (yf - ym) / ys, (yv - ym) / ys

    model = QCL()
    rng = np.random.default_rng(seed)
    psi_f, psi_v, psi_te = model.encode(Xf_s), model.encode(Xv_s), model.encode(Xte_s)

    theta = rng.uniform(0, 2 * np.pi, model.P)
    w = rng.normal(0, 0.3, N_Q)
    b = np.array([0.0])
    params = [theta, w, b]
    m = [np.zeros_like(p) for p in params]
    v = [np.zeros_like(p) for p in params]
    step = 0
    calls = 0
    best = {"val": np.inf, "state": None, "calls": 0, "epoch": 0}
    stale = 0
    hist = []
    Nf = len(yf_s)

    t0 = time.perf_counter()
    for epoch in range(1, MAX_EPOCHS + 1):
        perm = rng.permutation(Nf)
        for s in range(0, Nf, BATCH):
            ib = perm[s:s + BATCH]
            B = len(ib)
            psiN = model.forward(theta, psi_f[ib])
            z = model.zexp(psiN)
            pred = z @ w + b[0]
            r = 2.0 * (pred - yf_s[ib]) / B
            g = [model.grad_theta(theta, w, psiN, r), r @ z, np.array([r.sum()])]
            calls += B * (1 + 2 * model.P)
            step += 1
            for i, (p, gi) in enumerate(zip(params, g)):
                m[i] = 0.9 * m[i] + 0.1 * gi
                v[i] = 0.999 * v[i] + 0.001 * gi ** 2
                mh, vh = m[i] / (1 - 0.9 ** step), v[i] / (1 - 0.999 ** step)
                p -= LR * mh / (np.sqrt(vh) + 1e-8)
        # validation (exact expectations, one circuit per validation sample)
        zv = model.zexp(model.forward(theta, psi_v))
        val_mse = float(np.mean((zv @ w + b[0] - yv_s) ** 2))
        calls += len(yv_s)
        hist.append(val_mse)
        if val_mse < best["val"] - 1e-4:
            best = {"val": val_mse, "state": (theta.copy(), w.copy(), b.copy()), "calls": calls, "epoch": epoch}
            stale = 0
        else:
            stale += 1
            if stale >= PATIENCE:
                break
    t_train = time.perf_counter() - t0
    calls_total = calls
    theta, w, b = best["state"]

    def predict(psi, shots=None):
        ps = model.forward(theta, psi)
        z = model.zexp(ps) if shots is None else shot_zexp(model, ps, shots, rng)
        return (z @ w + b[0]) * ys + ym

    t0 = time.perf_counter()
    p_te_exact = predict(psi_te)
    t_inf_exact = time.perf_counter() - t0
    t0 = time.perf_counter()
    p_te_shots = predict(psi_te, INFER_SHOTS)
    t_inf_shots = time.perf_counter() - t0
    p_f = predict(psi_f)

    ridge = Ridge(alpha=1.0).fit(Xf_s, yf)
    out = {
        "dataset": name, "seed": seed,
        "qcl_exact": reg_metrics(yte, p_te_exact),
        "qcl_shots": reg_metrics(yte, p_te_shots),
        "qcl_train": reg_metrics(yf, p_f),
        "ridge": reg_metrics(yte, ridge.predict(Xte_s)),
        "mean_baseline": reg_metrics(yte, np.full_like(yte, yf.mean())),
        "cost": {
            "qubits": N_Q, "depth": model.depth, "params_circuit": model.P, "params_classical": N_Q + 1,
            "cz_gates": model.n_cz, "single_qubit_gates": model.n_1q,
            "epochs_run": len(hist), "best_epoch": best["epoch"],
            "calls_to_convergence": best["calls"], "calls_total_train": calls_total,
            "shots_per_circuit": TRAIN_SHOTS_ASSUMED,
            "train_shots_to_convergence": best["calls" ] * TRAIN_SHOTS_ASSUMED,
            "inference_circuits": len(yte), "inference_shots": len(yte) * INFER_SHOTS,
            "t_train_sim": t_train, "t_inf_exact": t_inf_exact, "t_inf_shots": t_inf_shots,
        },
    }
    return out


def self_check():
    model = QCL(n=6, layers=2)
    rng = np.random.default_rng(1)
    X = rng.uniform(-1, 1, (3, 6))
    theta = rng.uniform(0, 2 * np.pi, model.P)
    w = rng.normal(size=6)
    psi0 = model.encode(X)
    g_adj = model.grad_theta(theta, w, model.forward(theta, psi0), np.ones(3))
    E = lambda th: float(np.sum(model.zexp(model.forward(th, psi0)) @ w))
    g_ps = np.zeros(model.P)
    for k in range(model.P):
        tp, tm = theta.copy(), theta.copy()
        tp[k] += np.pi / 2
        tm[k] -= np.pi / 2
        g_ps[k] = (E(tp) - E(tm)) / 2
    norm = np.linalg.norm(model.forward(theta, psi0), axis=1)
    print("self-check: max |adjoint - parameter-shift| = %.2e ; max |norm-1| = %.2e"
          % (np.max(np.abs(g_adj - g_ps)), np.max(np.abs(norm - 1))))


def aggregate(results, path="qcl_report.md"):
    def fm(rows, getter, p=3):
        a = np.array([getter(r) for r in rows], float)
        return f"{a.mean():.{p}f}±{a.std():.{p}f}"

    out = []
    for ds in ["diabetes_D07", "friedman1_D08"]:
        rows = [r for r in results if r["dataset"] == ds]
        out.append(f"\n## {ds}  ({len(rows)} seeds, test partition)")
        out.append("| model | MAE | RMSE | R² | residual bias (pred−true) |")
        out.append("|---|---|---|---|---|")
        for label, key in [("QCL (exact expectations)", "qcl_exact"),
                           (f"QCL ({INFER_SHOTS}-shot inference)", "qcl_shots"),
                           ("Ridge (reference)", "ridge"), ("Predict train mean", "mean_baseline")]:
            out.append("| %s | %s | %s | %s | %s |" % (label,
                       fm(rows, lambda r: r[key]["mae"]), fm(rows, lambda r: r[key]["rmse"]),
                       fm(rows, lambda r: r[key]["r2"]), fm(rows, lambda r: r[key]["bias"])))
        out.append(f"\nQCL train-partition R² (exact): {fm(rows, lambda r: r['qcl_train']['r2'])}")
        c = lambda k: fm(rows, lambda r: r["cost"][k], 0)
        out.append("\nQuantum cost")
        out.append("| qubits | depth (est) | circuit params | CZ gates | epochs run | best epoch | circuit calls to convergence | total train calls | train shots to conv. (@1024/circ) | inference circuits | inference shots | t_train (sim) s | t_inf exact s | t_inf shots s |")
        out.append("|---|---|---|---|---|---|---|---|---|---|---|---|---|---|")
        r0 = rows[0]["cost"]
        out.append("| %d | %d | %d | %d | %s | %s | %s | %s | %s | %d | %d | %s | %s | %s |" % (
            r0["qubits"], r0["depth"], r0["params_circuit"], r0["cz_gates"],
            fm(rows, lambda r: r["cost"]["epochs_run"], 1), fm(rows, lambda r: r["cost"]["best_epoch"], 1),
            c("calls_to_convergence"), c("calls_total_train"), fm(rows, lambda r: r["cost"]["train_shots_to_convergence"], 0),
            r0["inference_circuits"], r0["inference_shots"],
            fm(rows, lambda r: r["cost"]["t_train_sim"], 1), fm(rows, lambda r: r["cost"]["t_inf_exact"], 4),
            fm(rows, lambda r: r["cost"]["t_inf_shots"], 3)))
    open(path, "w").write("\n".join(out))
    print("\n".join(out))


if __name__ == "__main__":
    # In interactive notebook sessions (like Google Colab), sys.argv contains kernel arguments (e.g., -f)
    # Instead of blindly parsing sys.argv[1], we check if a valid numeric seed count argument was passed.
    n_seeds = 5
    if len(sys.argv) > 1:
        try:
            n_seeds = int(sys.argv[1])
        except ValueError:
            pass

    self_check()
    results = []
    for ds in ["diabetes_D07", "friedman1_D08"]:
        for s in range(n_seeds):
            t0 = time.perf_counter()
            results.append(run_once(ds, s))
            print(f"{ds} seed {s}: R2 exact={results[-1]['qcl_exact']['r2']:.3f} "
                  f"epochs={results[-1]['cost']['epochs_run']}  ({time.perf_counter() - t0:.0f}s)", flush=True)
    json.dump(results, open("qcl_results.json", "w"), indent=1)
    aggregate(results)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Set plotting style
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# We will extract predictions vs actuals for the final seed of both datasets
# to demonstrate the performance visually.
for ax, ds_name in zip(axes, ["diabetes_D07", "friedman1_D08"]):
    # Re-run the final seed prediction for visualization
    X, y = get_data(ds_name, seed=4)
    Xtmp, Xte, ytmp, yte = train_test_split(X, y, test_size=0.3, random_state=4)
    Xf, Xv, yf, yv = train_test_split(Xtmp, ytmp, test_size=0.15, random_state=4)
    sc = MinMaxScaler((-1, 1)).fit(Xf)
    f = lambda A: np.clip(sc.transform(A), -1, 1)
    Xte_s = f(Xte)

    # Get QCL model parameters for seed 4
    ds_results = [r for r in results if r["dataset"] == ds_name and r["seed"] == 4]
    if ds_results:
        # We can plot predicted vs true targets
        # Let's run a quick inference to get individual predictions
        # using the last state to plot a scatter plot
        model = QCL()
        psi_te = model.encode(Xte_s)
        # Since run_once doesn't return the raw predictions in results,
        # we will plot a comparison of the metrics across models instead.

        models = ['QCL (Exact)', 'QCL (Shots)', 'Ridge']
        maes = [
            np.mean([r['qcl_exact']['mae'] for r in results if r['dataset'] == ds_name]),
            np.mean([r['qcl_shots']['mae'] for r in results if r['dataset'] == ds_name]),
            np.mean([r['ridge']['mae'] for r in results if r['dataset'] == ds_name])
        ]
        errors = [
            np.std([r['qcl_exact']['mae'] for r in results if r['dataset'] == ds_name]),
            np.std([r['qcl_shots']['mae'] for r in results if r['dataset'] == ds_name]),
            np.std([r['ridge']['mae'] for r in results if r['dataset'] == ds_name])
        ]

        ax.bar(models, maes, yerr=errors, capsize=10, color=['#4C72B0', '#55A868', '#C44E52'], alpha=0.8)
        ax.set_title(f"Model MAE Comparison on {ds_name} (Mean ± Std)")
        ax.set_ylabel("Mean Absolute Error (Lower is better)")

plt.tight_layout()
plt.show()
```
