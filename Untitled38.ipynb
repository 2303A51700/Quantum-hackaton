```python
#!/usr/bin/env python3
"""
P12 - Quantum LSTM (QLSTM) for temporal prediction on generated data:
      D19 noisy sine sequences, D20 NARMA-10 sequences.

Model
-----
Hybrid quantum-classical LSTM. Each of the four gates (forget, input, candidate, output) is a
variational quantum circuit (VQC) sandwiched between two classical linear maps:

    v_t = [x_t, h_{t-1}]
    a   = W_in v_t                        (angles for the circuit)
    q   = <Z_i> of VQC(a; theta)          (n qubits, angle encoding RY(a_i), L variational layers
                                           of RY, RZ on every qubit followed by a CNOT ring)
    s   = W_out q + b                     -> sigmoid (f, i, o) or tanh (candidate g)
    c_t = f*c_{t-1} + i*g ;  h_t = o*tanh(c_t)        (cell update is classical)

Training is full BPTT. The VQC Jacobians (w.r.t. circuit parameters AND input angles) come from
the parameter-shift rule: 2(P + n) extra circuit evaluations per gate call. No autograd framework
is used, so this file needs only numpy, scipy, scikit-learn, matplotlib.

`--model classical` swaps each VQC for tanh(a), giving a classical LSTM of identical structure.

Data protocol (matches the assignment text)
-------------------------------------------
* Raw series is split chronologically (train / val / test) into DISJOINT index ranges.
* Windows (inputs + forecast targets) are built inside each segment only, so no raw observation is
  used by two splits. A programmatic assertion checks this.
* StandardScaler is fitted on the training segment only.

Simulator notes (state these in the report)
-------------------------------------------
* Circuits are simulated exactly with a NumPy statevector (shots=0 = exact expectations).
  `--shots S` adds binomial shot noise to every circuit output, including the shifted circuits.
* Depth is a logical depth of one VQC computed from the gate list, not a transpiled depth.

Usage
-----
    python qlstm.py --dataset both --model both --epochs 30 --seeds 3
    python qlstm.py --dataset sine --model qlstm --epochs 5 --seeds 1      # quick check
    python qlstm.py --dataset narma --shots 1024                          # finite-shot training
"""
import argparse
import json
import sys
import time

import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler

GATES = "figo"


# --------------------------------------------------------------------------- data
def gen_sine(L, rng, period=25.0, noise=0.1):
    t = np.arange(L)
    y = np.sin(2 * np.pi * t / period) + noise * rng.standard_normal(L)
    return y[:, None], 0          # features (L,1), index of target column


def gen_narma10(L, rng, washout=100):
    for _ in range(20):
        n = L + washout
        u = rng.uniform(0.0, 0.5, n)
        y = np.zeros(n)
        for t in range(9, n - 1):
            y[t + 1] = 0.3 * y[t] + 0.05 * y[t] * y[t - 9: t + 1].sum() + 1.5 * u[t - 9] * u[t] + 0.1
        if np.all(np.isfinite(y)) and np.abs(y).max() < 10:
            return np.stack([u[washout:], y[washout:]], axis=1), 1   # features [u, y], target = y
    raise RuntimeError("NARMA-10 diverged repeatedly")


def make_windows(F_seg, ycol, W, H, stride, offset):
    """Window i: inputs rows [i, i+W), targets y at rows [i+W, i+W+H). Returns X, Y, raw index sets."""
    Ls = len(F_seg)
    xs, ys, used = [], [], set()
    for i in range(0, Ls - W - H + 1, stride):
        xs.append(F_seg[i: i + W])
        ys.append(F_seg[i + W: i + W + H, ycol])
        used.update(range(offset + i, offset + i + W + H))
    return np.array(xs), np.array(ys), used


def prepare(dataset, args, seed):
    rng = np.random.default_rng(1000 + seed)
    F, ycol = gen_sine(args.length, rng) if dataset == "sine" else gen_narma10(args.length, rng)
    L = len(F)
    n_tr, n_va = int(0.7 * L), int(0.15 * L)
    segs = {"train": (0, n_tr), "val": (n_tr, n_tr + n_va), "test": (n_tr + n_va, L)}
    sc = StandardScaler().fit(F[: n_tr])                     # fit on training segment only
    Fs = sc.transform(F)
    out, used_sets = {}, {}
    for name, (a, b) in segs.items():
        stride = args.stride if name == "train" else 1
        X, Y, used = make_windows(Fs[a:b], ycol, args.window, args.horizon, stride, a)
        out[name] = (X, Y)
        used_sets[name] = used
    names = list(segs)
    for i in range(3):
        for j in range(i + 1, 3):
            assert not (used_sets[names[i]] & used_sets[names[j]]), "split leakage!"
    y_mean, y_std = sc.mean_[ycol], sc.scale_[ycol]
    return out, (y_mean, y_std), F.shape[1], ycol


# --------------------------------------------------------------------------- circuit simulator
def circuit_depth(n, L):
    """Logical depth of one VQC computed from its gate list (each gate occupies its qubits)."""
    t = [0] * n
    def g1(i): t[i] += 1
    def g2(i, j):
        m = max(t[i], t[j]) + 1
        t[i] = t[j] = m
    for i in range(n): g1(i)                                 # RY(angle) encoding
    for _ in range(L):
        for i in range(n): g1(i); g1(i)                      # RY, RZ
        for i in range(n): g2(i, (i + 1) % n)                # CNOT ring
    return max(t)


class VQC:
    def __init__(self, n, L, shots, rng):
        self.n, self.L, self.P = n, L, L * n * 2
        self.shots, self.rng = shots, rng
        idx = np.arange(2 ** n)
        bits = (idx[:, None] >> (n - 1 - np.arange(n))[None, :]) & 1     # qubit 0 = most significant
        self.zsign = 1.0 - 2.0 * bits
        f = idx.copy()
        for c, t in [(i, (i + 1) % n) for i in range(n)]:
            cb, tb = n - 1 - c, n - 1 - t
            f = np.where((f >> cb) & 1, f ^ (1 << tb), f)
        self.inv_perm = np.argsort(f)
        self.evals = 0

    @staticmethod
    def _ry(a):
        c, s = np.cos(a / 2), np.sin(a / 2)
        m = np.zeros(a.shape + (2, 2), dtype=complex)
        m[..., 0, 0], m[..., 0, 1], m[..., 1, 0], m[..., 1, 1] = c, -s, s, c
        return m

    @staticmethod
    def _rz(a):
        m = np.zeros(a.shape + (2, 2), dtype=complex)
        m[..., 0, 0], m[..., 1, 1] = np.exp(-0.5j * a), np.exp(0.5j * a)
        return m

    def _apply(self, psi, m, q):
        N = psi.shape[0]
        p4 = psi.reshape(N, 2 ** q, 2, 2 ** (self.n - q - 1))
        return np.einsum("nab,nlbr->nlar", m, p4).reshape(N, -1)

    def _run(self, a, th):
        N, n = a.shape
        psi = np.zeros((N, 2 ** n), dtype=complex)
        psi[:, 0] = 1.0
        for i in range(n):
            psi = self._apply(psi, self._ry(a[:, i]), i)
        th = th.reshape(N, self.L, n, 2)
        for l in range(self.L):
            for i in range(n):
                m = np.einsum("nab,nbc->nac", self._rz(th[:, l, i, 1]), self._ry(th[:, l, i, 0]))
                psi = self._apply(psi, m, i)
            psi = psi[:, self.inv_perm]
        q = (psi.real ** 2 + psi.imag ** 2) @ self.zsign
        if self.shots:
            p1 = np.clip((1 + q) / 2, 0, 1)
            q = 2.0 * self.rng.binomial(self.shots, p1) / self.shots - 1.0
        return q

    def __call__(self, a, th, grad):
        B, n = a.shape
        P = self.P
        if not grad:
            self.evals += B
            return self._run(a, np.repeat(th[None], B, 0)), None, None
        V = 1 + 2 * (P + n)
        a_all = np.repeat(a[None], V, 0)
        th_all = np.repeat(th[None], V, 0)
        for p in range(P):
            th_all[1 + 2 * p, p] += np.pi / 2
            th_all[2 + 2 * p, p] -= np.pi / 2
        for i in range(n):
            a_all[1 + 2 * P + 2 * i, :, i] += np.pi / 2
            a_all[2 + 2 * P + 2 * i, :, i] -= np.pi / 2
        out = self._run(a_all.reshape(V * B, n), np.repeat(th_all, B, 0)).reshape(V, B, n)
        self.evals += V * B
        Jth = (out[1: 1 + 2 * P: 2] - out[2: 2 + 2 * P: 2]) / 2.0                 # (P,B,n)
        Ja = (out[1 + 2 * P:: 2] - out[2 + 2 * P:: 2]) / 2.0                      # (n,B,n)
        return out[0], np.transpose(Jth, (1, 2, 0)), np.transpose(Ja, (1, 2, 0))  # (B,n,P),(B,n,n_in)


# --------------------------------------------------------------------------- model
sigmoid = lambda z: 1.0 / (1.0 + np.exp(-z))


class QLSTM:
    def __init__(self, d_x, n, L, H_f, kind, shots, rng):
        self.d_x, self.n, self.kind = d_x, n, kind
        self.vqc = VQC(n, L, shots, rng) if kind == "qlstm" else None
        d = d_x + n
        p = {}
        for g in GATES:
            p[f"Win_{g}"] = rng.normal(0, 0.5 / np.sqrt(d), (n, d))
            p[f"Wout_{g}"] = rng.normal(0, 1 / np.sqrt(n), (n, n))
            p[f"b_{g}"] = np.full(n, 1.0 if g == "f" else 0.0)
            if self.vqc:
                p[f"th_{g}"] = rng.uniform(-np.pi, np.pi, self.vqc.P) * 0.5
        p["Wy"] = rng.normal(0, 1 / np.sqrt(n), (H_f, n))
        p["by"] = np.zeros(H_f)
        self.p = p

    def n_params(self):
        return sum(v.size for v in self.p.values())

    def _gate(self, g, v, grad):
        p = self.p
        a = v @ p[f"Win_{g}"].T
        if self.vqc:
            q, Jth, Ja = self.vqc(a, p[f"th_{g}"], grad)
        else:
            q = np.tanh(a)
            Jth = None
            Ja = None
            if grad:
                Ja = np.zeros(a.shape[:1] + (self.n, self.n))
                for k in range(self.n):
                    Ja[:, k, k] = 1 - q[:, k] ** 2
        s = q @ p[f"Wout_{g}"].T + p[f"b_{g}"]
        return a, q, Jth, Ja, (np.tanh(s) if g == "g" else sigmoid(s))

    def forward(self, X, grad=False):
        B, T, _ = X.shape
        h = np.zeros((B, self.n))
        c = np.zeros((B, self.n))
        cache = []
        for t in range(T):
            v = np.concatenate([X[:, t], h], axis=1)
            gs = {g: self._gate(g, v, grad) for g in GATES}
            f, i, gg, o = (gs[g][4] for g in GATES)
            c_new = f * c + i * gg
            tc = np.tanh(c_new)
            h_new = o * tc
            if grad:
                cache.append((v, gs, c, c_new, tc))
            h, c = h_new, c_new
        yhat = h @ self.p["Wy"].T + self.p["by"]
        return yhat, (cache, h)

    def backward(self, X, Y, yhat, cache_h):
        cache, h_last = cache_h
        p = self.p
        B, Hf = Y.shape
        grads = {k: np.zeros_like(v) for k, v in p.items()}
        dy = 2.0 * (yhat - Y) / (B * Hf)
        grads["Wy"] += dy.T @ h_last
        grads["by"] += dy.sum(0)
        dh = dy @ p["Wy"]
        dc = np.zeros_like(dh)
        for (v, gs, c_prev, c, tc) in reversed(cache):
            f, i, gg, o = (gs[g][4] for g in GATES)
            do = dh * tc
            dc = dc + dh * o * (1 - tc ** 2)
            dgate = {"f": dc * c_prev, "i": dc * gg, "g": dc * i, "o": do}
            dc_prev = dc * f
            dv = np.zeros_like(v)
            for g in GATES:
                a, q, Jth, Ja, act = gs[g]
                ds = dgate[g] * ((1 - act ** 2) if g == "g" else act * (1 - act))
                grads[f"Wout_{g}"] += ds.T @ q
                grads[f"b_{g}"] += ds.sum(0)
                dq = ds @ p[f"Wout_{g}"]
                if self.vqc:
                    grads[f"th_{g}"] += np.einsum("bo,bop->p", dq, Jth)
                da = np.einsum("bo,boi->bi", dq, Ja)
                grads[f"Win_{g}"] += da.T @ v
                dv += da @ p[f"Win_{g}"]
            dh = dv[:, self.d_x:]
            dc = dc_prev
        return grads

    def predict(self, X, batch=256):
        return np.concatenate([self.forward(X[i: i + batch], grad=False)[0] for i in range(0, len(X), batch)])


class Adam:
    def __init__(self, params, lr):
        self.lr, self.t = lr, 0
        self.m = {k: np.zeros_like(v) for k, v in params.items()}
        self.v = {k: np.zeros_like(v) for k, v in params.items()}

    def step(self, params, grads):
        self.t += 1
        for k in params:
            self.m[k] = 0.9 * self.m[k] + 0.1 * grads[k]
            self.v[k] = 0.999 * self.v[k] + 0.001 * grads[k] ** 2
            mh, vh = self.m[k] / (1 - 0.9 ** self.t), self.v[k] / (1 - 0.999 ** self.t)
            params[k] -= self.lr * mh / (np.sqrt(vh) + 1e-8)


# --------------------------------------------------------------------------- metrics
def regression_metrics(y_true, y_pred):
    yt, yp = y_true.ravel(), y_pred.ravel()
    return {"mae": mean_absolute_error(yt, yp), "rmse": float(np.sqrt(mean_squared_error(yt, yp))),
            "r2": r2_score(yt, yp), "bias": float(np.mean(yp - yt))}


def per_horizon(y_true, y_pred):
    return [{"h": h + 1, "mae": mean_absolute_error(y_true[:, h], y_pred[:, h]),
             "rmse": float(np.sqrt(mean_squared_error(y_true[:, h], y_pred[:, h]))),
             "r2": r2_score(y_true[:, h], y_pred[:, h])} for h in range(y_true.shape[1])]


# --------------------------------------------------------------------------- one run
def run_one(dataset, kind, seed, args):
    data, (ym, ys), d_x, ycol = prepare(dataset, args, seed)
    (Xtr, Ytr), (Xva, Yva), (Xte, Yte) = data["train"], data["val"], data["test"]
    rng = np.random.default_rng(seed)
    model = QLSTM(d_x, args.qubits, args.layers, args.horizon, kind, args.shots, rng)
    opt = Adam(model.p, args.lr)
    hist = {"train_loss": [], "val_loss": [], "grad_norm": []}
    best, best_p, bad = np.inf, None, 0
    t0 = time.perf_counter()
    diverged = False
    for ep in range(args.epochs):
        order = rng.permutation(len(Xtr))
        losses, gnorms = [], []
        for i in range(0, len(order), args.batch):
            idx = order[i: i + args.batch]
            yhat, ch = model.forward(Xtr[idx], grad=True)
            losses.append(float(np.mean((yhat - Ytr[idx]) ** 2)))
            gr = model.backward(Xtr[idx], Ytr[idx], yhat, ch)
            gn = float(np.sqrt(sum((g ** 2).sum() for g in gr.values())))
            gnorms.append(gn)
            if gn > args.clip:
                gr = {k: g * args.clip / gn for k, g in gr.items()}
            opt.step(model.p, gr)
        vl = float(np.mean((model.predict(Xva) - Yva) ** 2))
        hist["train_loss"].append(float(np.mean(losses)))
        hist["val_loss"].append(vl)
        hist["grad_norm"].append(float(np.mean(gnorms)))
        if not np.isfinite(vl):
            diverged = True
            break
        if vl < best - 1e-6:
            best, best_p, bad = vl, {k: v.copy() for k, v in model.p.items()}, 0
        else:
            bad += 1
            if bad >= args.patience:
                break
    train_time = time.perf_counter() - t0
    train_evals = model.vqc.evals if model.vqc else 0
    if best_p is not None:
        model.p = best_p
    if model.vqc:
        model.vqc.evals = 0
    t1 = time.perf_counter()
    pred = model.predict(Xte)
    inf_time = time.perf_counter() - t1
    inf_evals = model.vqc.evals if model.vqc else 0

    inv = lambda z: z * ys + ym
    yt, yp = inv(Yte), inv(pred)
    res = {"dataset": dataset, "model": kind, "seed": seed, "diverged": diverged,
           "epochs_run": len(hist["train_loss"]), "best_val_loss": best,
           "n_params": model.n_params(), "n_test": len(Xte),
           **regression_metrics(yt, yp), "per_horizon": per_horizon(yt, yp), "history": hist,
           "train_time_s": train_time, "inference_time_s": inf_time,
           "inference_time_per_sample_s": inf_time / len(Xte)}
    # persistence baseline: repeat the last observed y of each window
    last_y = inv(Xte[:, -1, ycol])[:, None].repeat(yt.shape[1], 1)
    res["persistence"] = regression_metrics(yt, last_y)
    gn = np.array(hist["grad_norm"])
    tl = np.array(hist["train_loss"])
    res["stability"] = {"max_grad_norm": float(gn.max()), "mean_grad_norm": float(gn.mean()),
                        "loss_spikes": int(np.sum(tl[1:] > 1.1 * tl[:-1])) if len(tl) > 1 else 0,
                        "final_train_loss": float(tl[-1])}
    if kind == "qlstm":
        V = 1 + 2 * (model.vqc.P + args.qubits)
        d = circuit_depth(args.qubits, args.layers)
        res["cost"] = {"qubits_per_gate_circuit": args.qubits, "qubits_if_gates_parallel": 4 * args.qubits,
                       "params_per_vqc": model.vqc.P, "vqc_logical_depth": d,
                       "depth_per_time_step_sequential_gates": 4 * d,
                       "depth_full_window_sequential": 4 * d * args.window,
                       "shots_per_circuit": args.shots if args.shots else "analytic (exact)",
                       "circuit_evals_per_gate_call_with_grad": V,
                       "circuit_evals_training": train_evals, "circuit_evals_inference": inf_evals,
                       "total_shots_training": train_evals * args.shots,
                       "total_shots_inference": inf_evals * args.shots}
    return res


def summarize(results, args):
    import matplotlib
    matplotlib.use("Agg")
    import matplotlib.pyplot as plt
    groups = {}
    for r in results:
        groups.setdefault((r["dataset"], r["model"]), []).append(r)
    print()
    for (ds, mk), rs in groups.items():
        print(f"== {ds.upper()} / {mk}  (seeds={len(rs)}, params={rs[0]['n_params']}, "
              f"diverged runs={sum(r['diverged'] for r in rs)}) ==")
        for k in ["mae", "rmse", "r2", "bias", "train_time_s", "inference_time_per_sample_s", "epochs_run"]:
            v = np.array([r[k] for r in rs], dtype=float)
            print(f"  {k:30s} {v.mean():.4f} +/- {v.std():.4f}")
        pm = np.array([r["persistence"]["mae"] for r in rs])
        print(f"  {"persistence baseline MAE":30s} {pm.mean():.4f} +/- {pm.std():.4f}")
        st = [r["stability"] for r in rs]
        print(f"  stability: max grad norm {max(s["max_grad_norm"] for s in st):.3f}, "
              f"loss spikes/run {np.mean([s["loss_spikes"] for s in st]):.1f}")
        ph = np.array([[h["mae"] for h in r["per_horizon"]] for r in rs]).mean(0)
        print("  MAE by horizon:", np.round(ph, 4).tolist())
        if "cost" in rs[0]:
            print("  cost:", rs[0]["cost"])
    ds_list = sorted({r["dataset"] for r in results})
    fig, axes = plt.subplots(2, len(ds_list), figsize=(5.5 * len(ds_list), 7), squeeze=False)
    for j, ds in enumerate(ds_list):
        for (d, mk), rs in groups.items():
            if d != ds:
                continue
            ph = np.array([[h["mae"] for h in r["per_horizon"]] for r in rs])
            axes[0, j].errorbar(np.arange(1, ph.shape[1] + 1), ph.mean(0), ph.std(0), marker="o", label=mk, capsize=3)
            for r in rs:
                axes[1, j].plot(r["history"]["val_loss"], alpha=0.6, label=f"{mk} s{r["seed"]}")
        axes[0, j].set(title=f"{ds}: test MAE vs forecast horizon", xlabel="horizon h", ylabel="MAE")
        axes[0, j].legend()
        axes[1, j].set(title=f"{ds}: validation loss per epoch", xlabel="epoch", ylabel="MSE (scaled)", yscale="log")
        axes[1, j].legend(fontsize=6)
    fig.tight_layout()
    fig.savefig(args.plot, dpi=130)
    print(f"saved {args.plot}")


def main():
    ap = argparse.ArgumentParser()
    ap.add_argument("--dataset", choices=["sine", "narma", "both"], default="both")
    ap.add_argument("--model", choices=["qlstm", "classical", "both"], default="both")
    ap.add_argument("--length", type=int, default=600, help="raw series length")
    ap.add_argument("--window", type=int, default=10)
    ap.add_argument("--horizon", type=int, default=5)
    ap.add_argument("--stride", type=int, default=3, help="stride for training windows")
    ap.add_argument("--qubits", type=int, default=4, help="qubits per gate circuit (= hidden size)")
    ap.add_argument("--layers", type=int, default=2)
    ap.add_argument("--shots", type=int, default=0, help="0 = exact expectations")
    ap.add_argument("--epochs", type=int, default=30)
    ap.add_argument("--batch", type=int, default=16)
    ap.add_argument("--lr", type=float, default=0.02)
    ap.add_argument("--clip", type=float, default=5.0)
    ap.add_argument("--patience", type=int, default=8)
    ap.add_argument("--seeds", type=int, default=3)
    ap.add_argument("--out", default="qlstm_results.json")
    ap.add_argument("--plot", default="qlstm_results.png")

    # Fix for running inside Jupyter/Colab notebooks:
    if len(sys.argv) > 1 and sys.argv[1] == "-f":
        args = ap.parse_args([])
    else:
        args = ap.parse_args()

    datasets = ["sine", "narma"] if args.dataset == "both" else [args.dataset]
    kinds = ["qlstm", "classical"] if args.model == "both" else [args.model]
    results = []
    for ds in datasets:
        for kind in kinds:
            for s in range(args.seeds):
                t = time.perf_counter()
                r = run_one(ds, kind, s, args)
                print(f"[{ds}/{kind}/seed {s}] MAE={r["mae"]:.4f} RMSE={r["rmse"]:.4f} R2={r["r2"]:.3f} "
                      f"epochs={r["epochs_run"]} ({time.perf_counter() - t:.0f}s)", flush=True)
                results.append(r)
    with open(args.out, "w") as f:
        json.dump(results, f, indent=1)
    summarize(results, args)


if __name__ == "__main__":
    main()
```
